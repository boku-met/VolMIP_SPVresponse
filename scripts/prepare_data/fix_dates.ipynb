```python
import glob
import numpy as np
import pandas as pd
import xarray as xr
from metpy.interpolate import log_interpolate_1d
from metpy.units import units
import matplotlib.pyplot as plt

time_coder = xr.coders.CFDatetimeCoder(use_cftime=True)
```

```python
models = ['CanESM5', 'GISS-E2-1-G','MPI-ESM1-2-LR', 'MIROC-ES2L', 'UKESM1-0-LL']
```

```python
# CanESM5
model = models[0]

for path in glob.glob('/sto4/data/Intermediate/VolMIP/pinatubo-full/' + model + '/ta*.nc'):
    ds = xr.open_dataset(path, decode_times=time_coder)
    ds['time'] = ds.indexes['time'].to_datetimeindex(time_unit='ns', unsafe=True).map(lambda t: t.replace(day=1, hour=0))
    ds.to_netcdf(path.split('.')[0] + '_fixed_dates.nc')


for path in glob.glob('/sto4/data/Intermediate/VolMIP/pinatubo-full/' + model + '/ua*.nc'):
    ds = xr.open_dataset(path, decode_times=time_coder)
    ds['time'] = ds.indexes['time'].to_datetimeindex(time_unit='ns', unsafe=True).map(lambda t: t.replace(day=1, hour=0))
    ds.to_netcdf(path.split('.')[0] + '_fixed_dates.nc')
```
