```python
import glob
import xarray as xr
import pandas as pd

time_coder = xr.coders.CFDatetimeCoder(use_cftime=True)
```

```python
models = ['CanESM5', 'GISS-E2-1-G', 'MPI-ESM1-2-LR', 'MIROC-ES2L', 'UKESM1-0-LL']
```

```python
# CanESM5
model = models[0]

for path in glob.glob('/sto4/data/Intermediate/VolMIP/pinatubo-full/' + model + '/utendepfd*.nc'):
    ds = xr.open_dataset(path, decode_times=time_coder)
    ds['time'] = ds.indexes['time'].to_datetimeindex(time_unit='ns', unsafe=True).map(lambda t: t.replace(day=1, hour=0))
    ds.to_netcdf(path.split('.')[0] + '_fixed_dates.nc')


# for path in glob.glob('/sto4/data/Intermediate/VolMIP/pinatubo-full/' + model + '/ENSO*.nc'):
#     ds = xr.open_dataset(path, decode_times=time_coder)
#     ds['time'] = ds.indexes['time'].to_datetimeindex(time_unit='ns', unsafe=True).map(lambda t: t.replace(day=1, hour=0))
#     ds.to_netcdf(path.split('.')[0] + '_fixed_dates.nc')
```
